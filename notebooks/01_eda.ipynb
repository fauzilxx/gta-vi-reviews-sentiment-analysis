Statistik dihitung dari CSV lokal. Asal yang diberikan pemilik proyek: https://www.kaggle.com/datasets/ektarr/gta-vi-user-reviews/data. Versi, lisensi, dan arti review_id belum terverifikasi independen.


```python
from pathlib import Path
import sys
import json
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'PRD.md').exists():
    ROOT = ROOT.parent
if not (ROOT / 'PRD.md').exists():
    raise RuntimeError('Buka notebook dari root proyek atau direktori notebooks')
sys.path.insert(0, str(ROOT))
from src.audit import describe_dataset
```

```python
raw = pd.read_csv(ROOT / 'data/raw/gta_vi_reviews.csv')
processed = pd.read_csv(ROOT / 'data/processed/processed_reviews.csv')
print(json.dumps({'raw': describe_dataset(raw), 'processed': describe_dataset(processed)}, indent=2))
```

```python
print('Panjang teks bersih (karakter):')
print(processed.clean_text.str.len().describe().to_string())
print('Distribusi label (%):')
print((processed.sentiment.value_counts(normalize=True) * 100).round(3).to_string())
```
